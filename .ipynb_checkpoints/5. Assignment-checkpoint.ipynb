## Question 1.

```python
class Vehicle:
    def display(self):
        print("This is a Vehicle")

class Car(Vehicle):
    def display(self):
        print("This is a Car")

class Bike(Vehicle):
    def display(self):
        print("This is a Bike")

N = int(input().strip())

if N == 2:
    obj = Bike()
elif N == 4:
    obj = Car()
else:
    obj = Vehicle()

obj.display()
```

## Question 3.

```python
class Student:
    def read(self):
        self.name = input().strip()
        self.marks = list(map(int, input().split()))
    
    def read_grade(self):
        self.grade = input().strip()

class Percentage(Student):
    def calculate_percentage(self):
        self.percentage = sum(self.marks) / 5

class Grade(Percentage):
    def get_grade(self):
        return self.grade

class Report(Grade):
    def show_report(self):
        print(self.name)
        print(*self.marks)
        print(f"{self.percentage:.2f}% {self.grade}")
```

```python
T = int(input())
for _ in range(T):
    obj = Report()
    obj.read()
    obj.read_grade()
    obj.calculate_percentage()
    obj.show_report()
```

## Question 4.

```python
import math

class Shapes:
    def read_rectangle(self):
        self.length, self.breadth = map(float, input().split())

    def read_square(self):
        self.side = float(input())

    def read_circle(self):
        self.radius = float(input())

    def read_rhombus(self):
        self.d1, self.d2 = map(float, input().split())

class Area(Shapes):
    def area_rectangle(self):
        return self.length * self.breadth

    def area_square(self):
        return self.side ** 2

    def area_circle(self):
        return 3.14 * (self.radius ** 2)

    def area_rhombus(self):
        return (self.d1 * self.d2) / 2


class Perimeter(Area):
    def perimeter_rectangle(self):
        return 2 * (self.length + self.breadth)

    def perimeter_square(self):
        return 4 * self.side

    def perimeter_circle(self):
        return 2 * 3.14 * self.radius

    def perimeter_rhombus(self):

        s = math.sqrt((self.d1 / 2) ** 2 + (self.d2 / 2) ** 2)
        return 4 * s

s = int(input())

for i in range(s):
    obj = Perimeter()

    obj.read_rectangle()
    obj.read_square()
    obj.read_circle()
    obj.read_rhombus()

    print(f"{int(obj.perimeter_rectangle())} {int(obj.area_rectangle())}")

    print(f"{int(obj.perimeter_square())} {int(obj.area_square())}")

    p_c = obj.perimeter_circle()
    a_c = obj.area_circle()
    print(f"{p_c:.2f} {a_c:.2f}")

    p_r = obj.perimeter_rhombus()
    a_r = obj.area_rhombus()
    print(f"{p_r:.2f} {a_r:.2f}")
```

## Question 5.

```python
class Base:
    def read(self):
        self.a1, self.b1 = map(int, input().split())
        self.a2, self.b2 = map(int, input().split())

class Sum(Base):
    def summation(self):
        return self.a1 + self.b1


class Product(Base):
    def multiplication(self):
        return self.a2 * self.b2


class Child(Sum, Product):
    def display(self):
        print(f"{self.a1} {self.b1}")
        print(self.summation())
        print(f"{self.a2} {self.b2}")
        print(self.multiplication())

s = int(input())
for i in range(s):
    obj = Child()
    obj.read()
    obj.display()
```

## Question 6.

```python
def sum_of_digits(n):
    return sum(int(d) for d in str(n))

N, Q = map(int, input().split())
A = list(map(int, input().split()))

for i in range(Q):
    i = int(input()) - 1 
    ans = -1
    sum_i = sum_of_digits(A[i])
    
    for j in range(i + 1, N):
        if A[i] < A[j] and sum_i > sum_of_digits(A[j]):
            ans = j + 1
            break
    print(ans)
```

## Question 7.

```python
N = int(input())
A = list(map(int, input().split()))

stack = []
G = []

for num in A:
    while stack and stack[-1] >= num:
        stack.pop()
    
    if not stack:
        G.append(-1)
    else:
        G.append(stack[-1])
    
    stack.append(num)

print(*G)
```

## Question 8.

```python
from collections import Counter

a = int(input())
for i in range(a
              ):
    N = int(input())
    A = list(map(int, input().split()))
    
    freq = Counter(A)
    res = [-1] * N
    stack = []
    
    for i in range(N - 1, -1, -1):
        while stack and freq[stack[-1]] <= freq[A[i]]:
            stack.pop()
        
        if stack:
            res[i] = stack[-1]
        else:
            res[i] = -1
        
        stack.append(A[i])
    
    print(*res)
```
