## Question 1.


```python
def batting_points(player):
    points = 0
    runs = player['runs']
    balls = player['balls']
    fours = player['4']
    sixes = player['6']
    field = player['field']

    points += runs // 2

    if runs >= 100:
        points += 10
    elif runs >= 50:
        points += 5

    sr = (runs / balls) * 100
    if 80 <= sr <= 100:
        points += 2
    elif sr > 100:
        points += 6 

    points += (fours * 1) + (sixes * 2)
    points += field * 10
    return points


def bowling_points(player):
    points = 0
    wkts = player['wkts']
    overs = player['overs']
    runs_given = player['runs']
    field = player['field']

    points += wkts * 10

    if wkts >= 5:
        points += 10
    elif wkts >= 3:
        points += 5

    eco = runs_given / overs
    if 3.5 <= eco <= 4.5:
        points += 4
    elif 2 <= eco < 3.5:
        points += 7
    elif eco < 2:
        points += 10

    points += field * 10

    return points
    
p1={'name':'Virat Kohli', 'role':'bat', 'runs':112, '4':10, '6':0, 'balls':119, 'field':0}
p2={'name':'du Plessis', 'role':'bat', 'runs':120, '4':11, '6':2, 'balls':112, 'field':0}
p3={'name':'Bhuvneshwar Kumar', 'role':'bowl', 'wkts':1, 'overs':10, 'runs':71, 'field':1}
p4={'name':'Yuzvendra Chahal', 'role':'bowl', 'wkts':2, 'overs':10, 'runs':45, 'field':0}
p5={'name':'Kuldeep Yadav', 'role':'bowl', 'wkts':3, 'overs':10, 'runs':34, 'field':0}

players = [p1, p2, p3, p4, p5]
scores = []

for p in players:
    if p['role'] == 'bat':
        score = batting_points(p)
        scores.append({'name': p['name'], 'batscore': score})
    else:
        score = bowling_points(p)
        scores.append({'name': p['name'], 'bowlscore': score})
for i in scores:
    print(i)
    
top_player = max(scores, key=lambda x: list(x.values())[1])
print(f"Man of the Match is : {top_player}")
```

## Question 2.

```python
def minimize_transactions():
    N, M = map(int, input().split())
    balance = [0] * (N + 1) 

    for _ in range(M):
        Vi, Ui, Wi = map(int, input().split())
        balance[Vi] -= Wi
        balance[Ui] += Wi
    total = sum(b for b in balance if b > 0)
    print(total)
```

```python
a = minimize_transactions()
```

```python
b = minimize_transactions()
```

## Question 3.

```python
def total_coins():
    T = int(input())
    for _ in range(T):
        a, k, n = map(int, input().split())
        total = a + k * (n - 1)
        print(total)
```

## Question 4.

```python
def camel_to_snake():
    T = int(input())
    for _ in range(T):
        S = input().strip()
        result = S[0]
        for ch in S[1:]:
            if ch.isupper():
                result += '_' + ch.lower()
            else:
                result += ch
        print(result)
```

```python
camel_to_snake()
```

## Question 5.

```python
def guess():
    T = int(input())
    for _ in range(T):
        X, Y, Z = map(int, input().split())
        A, B = map(int, input().split())

        best_bet = 0
        best_win_prob = -1

        for bet in range(X + 1):
            win_prob = 0

            for alice_correct in [True, False]:
                for bob_correct in [True, False]:
                    my_score = X + bet if True else X - bet  
                    if True:
                        pass

            for you_correct in [True, False]:
                for alice_correct in [True, False]:
                    for bob_correct in [True, False]:
                        your_score = X + bet if you_correct else X - bet
                        alice_score = Y + A if alice_correct else Y - A
                        bob_score = Z + B if bob_correct else Z - B

                        if your_score > alice_score and your_score > bob_score:
                            win_prob += 1

            win_prob /= 8

            if win_prob > best_win_prob:
                best_win_prob = win_prob
                best_bet = bet

        print(best_bet)
```

```python
guess()
```

## Question 6.

```python
import math

def guess_number():
    s = input().strip()
    found = False

    for n in range(1, 101):
        sq = str(n * n)

        if len(sq) != len(s):
            continue

        match = True
        for a, b in zip(s, sq):
            if a != '*' and a != b:
                match = False
                break

        if match:
            print(n)
            found = True
            break

    if not found:
        print(0)
```

```python
guess_number()
```

## Question 7.

```python
def count_rectangles():
    T = int(input())
    for _ in range(T):
        S = int(input())
        n = int(S ** 0.5) 
        total = 0
        for i in range(1, n + 1):
            for j in range(1, n + 1):
                if i * j <= S:
                    total += 1
        print(total)
```

```python
count_rectangles()
```
