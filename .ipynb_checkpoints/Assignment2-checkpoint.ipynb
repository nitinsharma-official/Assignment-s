## Question 1.

```python
def minimumNumber(n, password):
    numbers = set("0123456789")
    lower_case = set("abcdefghijklmnopqrstuvwxyz")
    upper_case = set("ABCDEFGHIJKLMNOPQRSTUVWXYZ")
    special_characters = set("!@#$%^&*()-+")
    chars = set(password)  
    missing = 0
    if not (chars & numbers):
        missing += 1
    if not (chars & lower_case):
        missing += 1
    if not (chars & upper_case):
        missing += 1
    if not (chars & special_characters):
        missing += 1  
    return max(missing, 6 - n)
if __name__ == '__main__':
    n = int(input("Enter length of number"))
    password = input("Enter your password")
    print(minimumNumber(n, password))
```

```python
minimumNumber(n, password)
```

## Question 2.



```python
def countApplesAndOranges(s, t, a, b, apples, oranges):

    apple_count = 0
    for d in apples:
        if s <= a + d <= t:
            apple_count += 1

    orange_count = 0
    for d in oranges:
        if s <= b + d <= t:
            orange_count += 1

    print(f"the number of apples that fall on Sam's house : {apple_count}")
    print(f"the number of oranges that fall on Sam's house : {orange_count}")

if __name__ == '__main__':
    st = input().split()
    s = int(st[0])
    t = int(st[1])
    
    ab = input().split()
    a = int(ab[0])
    b = int(ab[1])

    mn = input().split()
    m = int(mn[0])
    n = int(mn[1])

    apples = list(map(int, input().rstrip().split()))
    oranges = list(map(int, input().rstrip().split()))
```

```python
countApplesAndOranges(s, t, a, b, apples, oranges)
```

## Question 3.


```python
def gameOfThrones(s):
    from collections import Counter
    freq = Counter(s)  
    odd_count = 0
    for count in freq.values():
        if count % 2 != 0:
            odd_count += 1
    
    if odd_count > 1:
        print("NO")
    else:
        print("YES")


if __name__ == '__main__':
    s = input().strip()
```

```python
gameOfThrones(s)
```

```python
gameOfThrones(s)
```

```python
gameOfThrones(s)
```

## Question 4.

```python
n = int(input())
for _ in range(n):
    SH, SM, EH, EM = map(int, input().split()) 
    start = SH * 60 + SM
    end = EH * 60 + EM
    duration = end - start 
    hours = duration // 60
    minutes = duration % 60 
    print(hours, minutes)
```

## Question 5.


```python
def num_pattern(n):
    n = int(input())
    size = 2 * n - 1
    center = n - 1
    i = 0
    while i < size:
        j = 0
        while j < size:
            layer = max(abs(i - center), abs(j - center))
            print(layer + 1, end=" ")
            j += 1
        print()
        i += 1
```

```python
num_pattern(2)
```

```python
num_pattern(5)
```

## Question 6.


```python
def print_Z_pattern(n):
    print('*' * n)
    i = n - 2
    while i > 0:
        print(" "* (i - 1) + " *")
        i -= 1
    print("*" * n)

n = int(input("Enter value of n 3 to 20.."))
if 2 < n <= 20:
    print_Z_pattern(n)
else:
    print("Invalid input..")
```

```python
print_Z_pattern(3)
```

```python
print_Z_pattern(4)
```

```python
print_Z_pattern(5)
```

## Question 7.


```python
def dayOfProgrammer(year):
    if year == 1918:
        return "26.09.1918"
    elif year < 1918:
        if year % 4 == 0:
            return f"12.09.{year}"
        else:
            return f"13.09.{year}"   
    else:
        if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
            return f"12.09.{year}"
        else:
            return f"13.09.{year}"
if __name__ == '__main__':
    year = int(input().strip())
```

```python
dayOfProgrammer(2017)
```

```python
dayOfProgrammer(2016)
```

```python
dayOfProgrammer(1800)
```
