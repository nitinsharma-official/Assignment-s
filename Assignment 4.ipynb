## Question 1.

```python
def chocolateFeast(n, c, m):
    chocolates = n // c
    wrappers = chocolates

    while wrappers >= m:
        new_chocolates = wrappers // m
        chocolates += new_chocolates
        wrappers = wrappers % m + new_chocolates

    return chocolates
t = int(input())   
for i in range(t):
    n, c, m = map(int, input().split())
    print(chocolateFeast(n, c, m))
```

## Question 2.

```python
def howManyGames(p, d, m, s):
    games = 0
    current_price = p

    while s >= current_price:
        s -= current_price
        games += 1
        current_price = max(current_price - d, m)  
    
    return games
```

```python
p, d, m, s = map(int, input().split())
print(howManyGames(p, d, m, s))
```

## Quesion 3.

```python
import sys
sys.setrecursionlimit(10000)

def can_reach_by_recursion(n):
    if n == 1:
        return True
    if n < 1:
        return False
    if n % 10 == 0:
        if can_reach_by_recursion(n // 10):
            return True
    if n % 20 == 0:
        if can_reach_by_recursion(n // 20):
            return True
    return False
```

```python
t = int(input().strip())
for _ in range(t):
    n = int(input().strip())
    print("Yes" if can_reach_by_recursion(n) else "No")
```

## Question 4.

```python
def find_winner(members, song, index, song_index):
    if len(members) == 1:
        return members[0]
    lyric = song[song_index % len(song)]    
    if lyric == 'x':
        next_index = (index + 1) % len(members)
        return find_winner(members, song, next_index, song_index + 1)
    elif lyric == 'y':
        members.pop(index)
        next_index = index % len(members)
        return find_winner(members, song, next_index, song_index + 1)

n = int(input())
song = input().strip()
members = list(range(1, n + 1))
print(find_winner(members, song, 0, 0))
```

## Question 5.

```python
import time
def final_position(commands):
    x, y = 0, 0
    for c in commands:
        if c == 'L':
            x -= 1
        elif c == 'R':
            x += 1
        elif c == 'U':
            y += 1
        elif c == 'D':
            y -= 1
    return x, y
```

```python
def main():

    start_time = time.time()
    
    commands = input().strip()
    

    time.sleep(2)
    

    x, y = final_position(commands)
    

    end_time = time.time()
    
    time_taken = end_time - start_time

    print(f"{x} {y}")
    print(f"Time taken: {time_taken:.10f} sec")

if __name__ == "__main__":
    main()
```

## Question 6.

```python
import time

def fredo_game(A, obstacles):
    for i in range(len(obstacles)):
        if obstacles[i] == 1:
            A += 2
        else:
            A -= 1
        
        if A == 0:
            return f"No {i + 1}"
    return f"Yes {A}"
```

```python
def main():
    T = int(input().strip())
    for _ in range(T):
        A, N = map(int, input().split())
        obstacles = list(map(int, input().split()))
        
        start_time = time.time()
        
        time.sleep(2)
        
        result = fredo_game(A, obstacles)
        
        end_time = time.time()
        time_taken = end_time - start_time
        
        print(result)
        print(f"Time taken : {time_taken:.10f} sec")


if __name__ == "__main__":
    main()
```

## Questin 7.

```python
def pattern_decorator(func):
    """Decorator to print stars before and after the pattern."""
    def wrapper(n):
        print("*********************")
        func(n)
        print("*********************")
    return wrapper


@pattern_decorator
def print_pattern(n):
    """Prints the square numeric pattern for a given N."""
    size = 2 * n - 1
    for i in range(size):
        for j in range(size):
            val = n - min(i, j, size - i - 1, size - j - 1)
            print(val, end=" ")
        print()
```

```python
if __name__ == "__main__":
    n = int(input().strip())
    print_pattern(n)
```
